## Tratamento do dataset

### Tratamento da data

```python
df['data_editada'] = df['date_x'].str.strip()
df['data'] = pd.to_datetime(df['data_editada'], format='%m/%d/%Y')
df['data']
```

```python
df['orig_lang'] = df['orig_lang'].str.strip()
```

### Tratamento do elenco

```python
df['elenco_lista'] = df['crew'].str.split(', ')
df['elenco_lista']

for lista in df['elenco_lista']:
    for item in lista[::2]:
        print(item)

# Até aqui deu pra pegar os itens
```
