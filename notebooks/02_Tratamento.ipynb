```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/ClaudiaSobral/treinamento-dados-end-to-end/refs/heads/main/data/silver/imdb_movies_segmentado.csv')
```

```python
df['score'].min()
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
filtro_estrelas = [
    (df.score <=20.0),
    (df.score > 20.0) & (df.score <=40.0),
    (df.score > 40.0) & (df.score <=60.0),
    (df.score > 60.0) & (df.score <=80.0),
    (df.score <=10.0)
]

categoria_estrelas = ["1_estrelas", "2_estrelas", "3_estrelas", "4_estrelas", "5_estrelas" ]
```

```python
df['estrelas'] = np.select(filtro_estrelas, categoria_estrelas)
df.head()
```
