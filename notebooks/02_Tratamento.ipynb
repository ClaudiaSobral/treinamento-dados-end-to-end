```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/ClaudiaSobral/treinamento-dados-end-to-end/refs/heads/main/data/silver/imdb_movies_segmentado.csv')
```

```python
df['score'].max()
```

```python
df.info()
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
filtro_estrelas = [
    (df['score'] <=20.0),
    (df['score'] > 20.0) & (df['score'] <=40.0),
    (df['score'] > 40.0) & (df['score'] <=60.0),
    (df['score'] > 60.0) & (df['score'] <=80.0),
    (df['score'] <=100.0)
]

categoria_estrelas = ["1", "2", "3", "4", "5"]
```

```python
df['estrelas'] = np.select(filtro_estrelas, categoria_estrelas, default="0")
df.head()
```

```python
df = df.drop(columns=['elenco_lista', 'date_x'])
```

```python
df.to_csv('../data/golden/imdb_movies_tratado.csv', index=False)
```
